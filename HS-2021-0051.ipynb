```python
# Enable Tkinter in Jupyter Notebook
%gui tk

import tkinter as tk
from tkinter import messagebox
from datetime import datetime

def is_leap_year(year):
    """Check if a year is a leap year."""
    year = int(year)
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def get_correct_date(year, day_part):
    """Fix day count for leap years."""
    year = int(year)
    if not is_leap_year(year) and day_part > 59:
        day_part -= 1  # Adjust for non-leap years after Feb 28
    return datetime.strptime(f"{year}-{day_part}", "%Y-%j").strftime("%Y-%m-%d")

def extract_details():
    nic = nic_entry.get().strip()
    
    if len(nic) == 10 and nic[-1] in ['V', 'X']:  # Old NIC (10-digit)
        year = "19" + nic[:2]
        day_part = int(nic[2:5])
    elif len(nic) == 12 and nic.isdigit():  # New NIC (12-digit)
        year = nic[:4]
        day_part = int(nic[4:7])
    else:
        messagebox.showerror("Invalid NIC", "Please enter a valid NIC number.")
        return
    
    # Determine gender
    if day_part > 500:
        gender = "Female"
        day_part -= 500
    else:
        gender = "Male"
    
    try:
        birth_date = get_correct_date(year, day_part)
        result_label.config(text=f"📅 Birthdate: {birth_date}\n🧑 Gender: {gender}", fg="blue", font=("Arial", 12, "bold"))
    except ValueError:
        messagebox.showerror("Invalid NIC", "Invalid birthdate calculation.")

# Tkinter GUI
root = tk.Tk()
root.title("🇱🇰 Sri Lanka NIC Extractor")
root.geometry("400x250")

tk.Label(root, text="Enter NIC Number:", font=("Arial", 11)).pack(pady=5)
nic_entry = tk.Entry(root, font=("Arial", 11))
nic_entry.pack(pady=5)

tk.Button(root, text="Extract Details", font=("Arial", 11, "bold"), command=extract_details).pack(pady=5)
result_label = tk.Label(root, text="", font=("Arial", 11))
result_label.pack(pady=10)

root.mainloop()
```
